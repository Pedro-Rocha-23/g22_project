```python
import pandas as pd
```

```python
df = pd.read_csv("g22_Research_Grants.csv", sep = ';')
df.columns = df.columns.str.strip()
df['region'] = df['region'].astype(str).str.strip()
```

```python
df_university = df.loc[:,['university_id','name','creation_date','region']]
```

```python
df_university = df_university.drop_duplicates(subset='university_id', keep='first')
```

```python
df_grant = df.loc[:,['grant_id','title','category','director_id']]
df_grant = df_grant.rename(columns={'grant_id':'id'})
```

```python
df_grant = df_grant.drop_duplicates(subset='id', keep='first')
```

```python
df_lab = df.loc[:,['lab_id','extra_info','university_id']]
df_lab = df_lab.rename(columns={'lab_id':'id'})
```

```python
df_lab = df_lab.drop_duplicates(subset='id', keep='first')
```

```python
df_lab = df_lab.dropna()
```

```python
df_lab['id'] = df_lab['id'].astype(int)
```

```python
df_director = df.loc[:,['director_id','director_name','date_of_birth']]
df_director = df_director.rename(columns={'director_id':'id', 'date_of_birth':'dob'})
```

```python
df_director = df_director.drop_duplicates(subset='id', keep='first')
```

```python
df_region = df.loc[:,['region']]
```

```python
df_region = df_region.drop_duplicates(subset='region', keep='first')
df_region = df_region.reset_index(drop=True)
df_region.insert(0, 'id', range(1, len(df_region) + 1))
df_region = df_region.rename(columns={'region':'name'})
```

```python
df_region_aux = df_region.rename(columns={'id':'region_id', 'name':'region'})
df_university = df_university.merge(df_region_aux, on='region', how='left')
df_university = df_university.rename(columns={'university_id':'id'})
df_university = df_university.loc[:,['id','name','creation_date','region_id']]
```

```python
df_uni_grant = df.loc[:,['university_id','grant_id','amount','start_date']]
```

```python
df_uni_grant = df_uni_grant.drop_duplicates(subset=['university_id', 'grant_id'], keep='first')
```

```python
df_uni_grant.insert(0, 'id', range(1, len(df_uni_grant) + 1))
```

```python
df_userlogin = pd.DataFrame({
    'id': [1, 2],
    'user': ['root', 'user1'],
    'usergroup': ['admin', 'users'],
    'password': ['1234', '12345']
})
```

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine('sqlite:///g22_db.db')
```

```python
df_university.to_sql('University', con=engine, if_exists='replace', index=False)
```

```python
df_grant.to_sql('Grant', con=engine, if_exists='replace', index=False)
```

```python
df_lab.to_sql('Lab', con=engine, if_exists='replace', index=False)
```

```python
df_director.to_sql('Director', con=engine, if_exists='replace', index=False)
```

```python
df_uni_grant.to_sql('Uni_grant', con=engine, if_exists='replace', index=False)
```

```python
df_region.to_sql('Region', con=engine, if_exists='replace', index=False)
```

```python
df_userlogin.to_sql('Userlogin', con=engine, if_exists='replace', index=False)
```

```python
engine.dispose()
```
